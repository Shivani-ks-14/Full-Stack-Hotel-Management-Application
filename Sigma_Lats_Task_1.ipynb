```python
# percentage of missing values in each column
missing_pct = (df.isnull().sum(axis=0)/len(df))*100
print(missing_pct)
```

We observe that govt_debt_pct_gdp, gini_index, and poverty_headcount_190 contain a large percentage of missing values (>50%).
Here gini_index and poverty_headcount_190 are useful indicators in predicting HDI, poverty or development. We may use certain intelligent imputation methods like country wise imputation or interpolation over years to fill in the null values. But we end up with a huge amount of artificial data.  
So we shall drop these features and evaluate our model performance.


```python
# drop columns with more than 50% missing values
df = df.loc[:,df.isnull().mean()<0.5]
```

```python
# recheck percentage of missing values
missing_pct = (df.isnull().sum(axis=0)/len(df))*100
print(missing_pct)
```

Now the columns with very high missing values are removed. The dataset in much more manageable.
Now we shall handle the remaining missing values using imputation.

We will do the imputation in stages


*   Interpolation by country - to preserve yearly trends
*   country wise median - to preserve country characteristics
*   Region wise median - use region similarity
*   global median fall back - to remove remaining null values

```python
# sort the data properly
df = df.sort_values(['iso3_code','year'])
```

```python
# separate numerical and categorical columns
num_cols = df.select_dtypes(include = ['float64','int64']).columns
cat_cols = df.select_dtypes(include = 'object').columns
```

```python
num_cols
```

```python
cat_cols
```

We separate numerical and categorical columns because we impute numerical columns with median, and categorical columns with mode.
Here the categorical columns have no null values, so no action required.

```python
# interpolation by country
for col in num_cols:
  df[col] = df.groupby('iso3_code')[col]\
  .transform(lambda x: x.interpolate())
```

```python
# country wise median imputation
for col in num_cols:
  df[col] = df.groupby('iso3_code')[col]\
  .transform(lambda x: x.fillna(x.median()))
```

```python
# region wise median imputation
for col in num_cols:
  df[col] = df.groupby('region')[col]\
  .transform(lambda x: x.fillna(x.median()))
```

```python
# global median fall back
for col in num_cols:
  df[col] = df[col].fillna(df[col].median())
```

```python
# verify that there's no more null values
df.isnull().sum(axis=0)
```

Therefore we have handled all the null values
